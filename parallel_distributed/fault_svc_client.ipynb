````python
import requests
from tenacity import retry, stop_after_attempt, wait_exponential

# Service URL
svc_url = "http://127.0.0.1:8000/svc-5"

@retry(
    stop=stop_after_attempt(4),
    wait=wait_exponential(multiplier=1, min=1, max=10),
    reraise=True  # Throws the original exception if all attempts fail
)
def fetch_api_data(id: int = 0):
    # Return 1 if successful, otherwise raise an exception
    print("Querying external API... ({})".format(id))
     # GET request with automatic JSON parsing
    response = requests.get(svc_url)
    data = response.json()
    #print(data.get('message'))
    print("\tAPI response:", data.get('message'))
    if not data.get('message').startswith('SUCCESS'):
        #print("\tAPI call successful! Data: ", data.get('message'))
    #else:
        raise ConnectionError(" Service failed to respond")
````

````python
import time

num_requests = 100
num_failed = 0

t_start = time.perf_counter()

for i in range(num_requests):
    try:
        fetch_api_data(i)
    except ConnectionError as e:
        print(" Service failed completely after 4 attempts:")
        num_failed += 1

# Timespan
elapsed_time = time.perf_counter() - t_start
# Success rate
success_rate = (num_requests - num_failed) / num_requests

print(elapsed_time, success_rate)
````
