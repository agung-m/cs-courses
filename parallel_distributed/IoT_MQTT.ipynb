```python
import paho.mqtt.client as mqtt
import ssl
import time

# 1. HiveMQ Configuration
broker_url = "broker.hivemq.com"
port = 8884
username = ""
password = ""
topic = "wokwi-weather" # Change to your specific topic
topic_gas = "wokwi-gas"

# 2. Define Callback Functions
def on_connect(client, userdata, flags, reason_code, properties=None):
    print(f"Connected with result code: {reason_code}")
    # Subscribe upon connecting successfully
    client.subscribe(topic)
    client.subscribe(topic_gas)
    print(f"Subscribed to topic: {topic, topic_gas}")

def on_message(client, userdata, msg):
    print(f"Received message on {msg.topic}: {msg.payload.decode('utf-8')}")

# 3. Initialize the Client (using Paho-MQTT v2 API) and websockets
client = mqtt.Client(callback_api_version=mqtt.CallbackAPIVersion.VERSION2, transport="websockets")
#client.username_pw_set(username, password
client.username_pw_set(username, password)

# 4. Set up TLS for secure HiveMQ Cloud connection
client.tls_set(tls_version=ssl.PROTOCOL_TLS)

# 5. Assign Callbacks
client.on_connect = on_connect
client.on_message = on_message

# Instruct the client to use WebSockets path
client.ws_set_options(path="/mqtt")

# 6. Connect to the broker and start the network loop
try:
    client.connect(broker_url, port=port)
    # Use loop_start() in Colab so the cell doesn't freeze and allows you to stop it gracefully
    client.loop_start()

    # Keep the cell running to listen for messages
    print("Listening for messages... (Press the Stop button in Colab to exit)")
    while True:
        time.sleep(1)
except KeyboardInterrupt:
    print("\nDisconnecting...")
    client.loop_stop()
    client.disconnect()
```
