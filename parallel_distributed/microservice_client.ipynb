```python
import requests
from tenacity import retry, stop_after_attempt, wait_exponential

#weibull_1_url = "http://127.0.0.1:8000/svc-weibull-1"
svc_url = "http://127.0.0.1:8000/svc-norm"

@retry(
    stop=stop_after_attempt(4),
    wait=wait_exponential(multiplier=1, min=1, max=10),
    reraise=True  # Throws the original exception if all attempts fail
)
def fetch_api_data(id: int = 0):
    # Return 1 if successful, otherwise raise an exception
    print("Querying external API... ({})".format(id))
     # GET request with automatic JSON parsing
    response = requests.get(svc_url)
    data = response.json()
    #print(data.get('message'))
    #print("API response:", data.get('message'))
    if data.get('message').startswith('SUCCESS'):
        print("\tAPI call successful! Data: ", data.get('message'))
    else:
        raise ConnectionError(" Service failed to respond. Retrying...")
```
